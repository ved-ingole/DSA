```python
# Greatest Common Divisor & Lowest Common Factor
def gcd(a, b):
  if b==0:
    return a
  return gcd(b, a%b)
def lcm(a, b):
  return a*b//gcd(a, b)
print(lcm(15, 50))
print(gcd(15,50))
```

```python
def findPow(x, n):
    # Base case
    if n == 0:
        return 1

    # Recursive call
    a = findPow(x, n // 2)

    # If n is even
    if n % 2 == 0:
        return a * a

    # If n is odd
    else:
        return a * a * x


x = int(input("Enter x: "))
n = int(input("Enter n: "))

print("Answer:", findPow(x, n))
```

```python
# 6 Buy 1 sell 1
prices = [7, 1, 5, 3, 6, 4]

def maxProfit(prices):
    if not prices:
        return 0

    min_price = prices[0]
    profit = 0

    for i in range(1, len(prices)):
        curr_profit = prices[i] - min_price

        if curr_profit > profit:
            profit = curr_profit

        min_price = min(min_price, prices[i])

    return profit


print(maxProfit(prices))
```

```python
def maxProfit(prices):
    profit = 0
    n = len(prices)

    for i in range(n - 1):
        if prices[i + 1] > prices[i]:
            profit += prices[i + 1] - prices[i]

    return profit


prices = [7, 1, 5, 3, 6, 4]

print(maxProfit(prices))
```

```python
def spiralOrder(matrix):

    n = len(matrix)
    m = len(matrix[0])

    total = n * m
    ans = []
    c = 0

    colstart = 0
    rowstart = 0
    colend = m - 1
    rowend = n - 1

    while c < total:

        # rowstart, colstart -> colend
        for i in range(colstart, colend + 1):
            ans.append(matrix[rowstart][i])
            c += 1

        rowstart += 1

        # rowstart -> rowend, colend
        for i in range(rowstart, rowend + 1):
            ans.append(matrix[i][colend])
            c += 1

        colend -= 1

        # colend -> colstart, rowend
        for i in range(colend, colstart - 1, -1):
            ans.append(matrix[rowend][i])
            c += 1

        rowend -= 1

        # rowend -> rowstart, colstart
        for i in range(rowend, rowstart - 1, -1):
            ans.append(matrix[i][colstart])
            c += 1

        colstart += 1

    return ans


matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(spiralOrder(matrix),)
```
