```python
def findPow(x, n):
    # Base case
    if n == 0:
        return 1

    # Recursive call
    a = findPow(x, n // 2)

    # If n is even
    if n % 2 == 0:
        return a * a

    # If n is odd
    else:
        return a * a * x


x = int(input("Enter x: "))
n = int(input("Enter n: "))

print("Answer:", findPow(x, n))
```

```python
# 6 Buy 1 sell 1
prices = [7, 1, 5, 3, 6, 4]

def maxProfit(prices):
    if not prices:
        return 0

    min_price = prices[0]
    profit = 0

    for i in range(1, len(prices)):
        curr_profit = prices[i] - min_price

        if curr_profit > profit:
            profit = curr_profit

        min_price = min(min_price, prices[i])

    return profit


print(maxProfit(prices))
```

```python
def spiralOrder(matrix):

    n = len(matrix)
    m = len(matrix[0])

    total = n * m
    ans = []
    c = 0

    colstart = 0
    rowstart = 0
    colend = m - 1
    rowend = n - 1

    while c < total:

        # rowstart, colstart -> colend
        for i in range(colstart, colend + 1):
            ans.append(matrix[rowstart][i])
            c += 1

        rowstart += 1

        # rowstart -> rowend, colend
        for i in range(rowstart, rowend + 1):
            ans.append(matrix[i][colend])
            c += 1

        colend -= 1

        # colend -> colstart, rowend
        for i in range(colend, colstart - 1, -1):
            ans.append(matrix[rowend][i])
            c += 1

        rowend -= 1

        # rowend -> rowstart, colstart
        for i in range(rowend, rowstart - 1, -1):
            ans.append(matrix[i][colstart])
            c += 1

        colstart += 1

    return ans


matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(spiralOrder(matrix),)
```

## 1. Bubble Sort

```python
def sortArray(nums):
  n = len(nums)
  print(n)

  for i in range(n):
    isSwap = False

    for j in range(n-i-1):
      if nums[j] > nums[j+1]:

        #swap
        temp = nums[j]
        nums[j] = nums[j+1]
        nums[j+1] = temp

        isSwap = True

    if not isSwap:
      break
  return nums
print(sortArray([3,2,8,1,7,6,4]))
```

## 2. Insertion Sort

```python
def sortArray(nums):
  n = len(nums)

  for i in range(1,n):
    key = nums [i]
    j = i-1
    while j>=0 and nums [j]>key:
        nums [j+1]=nums [ j ]
        j -= 1
        nums [j+1] = key
  return nums
print(sortArray([3,2,8,1,7,6,4]))
```

## 3. Selection Sort

```python
def sortArray(nums):
  n = len(nums)
  for i in range(n):
    mn = nums[i]
    for j in range(i+1,n):
      if nums[j]<mn:
        mn = nums[j]
        ind = j
    temp = nums[i]
    nums[i] = nums[ind]
    nums[ind] = temp
  return nums
```

```python
# 1. Middle of the linked list
class SinglyNode:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next

def findMiddle(head):
    slow = head
    fast = head

    while fast != None and fast.next != None:
        slow = slow.next
        fast = fast.next.next

    return slow

# Create linked list
head = SinglyNode(1)
head.next = SinglyNode(2)
head.next.next = SinglyNode(3)
head.next.next.next = SinglyNode(4)
head.next.next.next.next = SinglyNode(5)
head.next.next.next.next.next = SinglyNode(6)

# Find middle
middle = findMiddle(head)

print("Middle node:", middle.data)
```

```python
# 4. Remove duplicates from sorted list
class SinglyNode:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next


a = SinglyNode(1)
b = SinglyNode(1)
c = SinglyNode(3)
d = SinglyNode(3)
e = SinglyNode(5)

a.next = b
b.next = c
c.next = d
d.next = e

head = a

curr = head

while curr != None and curr.next != None:
    if curr.data == curr.next.data:
        curr.next = curr.next.next
    else:
        curr = curr.next

# Print
curr = head

while curr != None:
    print(curr.data, end=" → ")
    curr = curr.next

print("None")
```

```python
#6. Rotate List

class SinglyNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


# Create nodes
a = SinglyNode(1)
b = SinglyNode(2)
c = SinglyNode(3)
d = SinglyNode(4)
e = SinglyNode(5)
f = SinglyNode(6)
g = SinglyNode(7)

# Connect nodes
a.next = b
b.next = c
c.next = d
d.next = e
e.next = f
f.next = g


k = 3


def rotateList(head):

    if head == None or head.next == None:
        return head

    # Find length
    l = 1
    last = head

    while last.next != None:
        last = last.next
        l += 1

    # Reduce k
    k1 = k % l

    if k1 == 0:
        return head

    # Find new last node
    curr = head

    for i in range(l - k1 - 1):
        curr = curr.next

    # Connect last node to head
    last.next = head

    # New head
    head = curr.next

    # Break the connection
    curr.next = None

    return head


# Rotate
head = rotateList(a)


# Print
curr = head

while curr != None:
    print(curr.val, end=" ")
    curr = curr.next
```

```python
class Stack:
  def __init__(self):
    self.st = []

  def push(self,x):
    self.st.append(x)

  def pop(self):
    if len(self.st) == 0:
      return -1

    x = self.st[-1]
    self.st.pop()
    return x

  def top(self):
    if len(self.st)  == 0:
      return -1
    return self.st[-1]

  def size(self):
    return len(self.st)

stack = Stack()
stack.push(7)
stack.push(8)
stack.push(9)

print(stack.pop())
print(stack.top())
print(stack.size())
```
